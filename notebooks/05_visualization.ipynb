Este notebook prepara y exporta las tablas finales curadas (`dashboard_ranking.csv`) para alimentar los mapas y dashboards del proyecto SIPTA utilizando `src.visualization.prepare_visualization`.

## 1. Configuración e Imports

```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve() if (Path('..') / 'src').exists() else Path('.').resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import src.visualization.prepare_visualization as viz

print(f"Raíz: {ROOT}")
print(f"Directorio curated: {viz.CURATED_DIR}")
```

## 2. Generación de Ranking Territorial para Dashboard

```python
# Ejemplo de ranking territorial sintético
df_ejemplo_ipt = pd.DataFrame({
    'localidad': ['BOSA', 'CIUDAD BOLIVAR', 'USME', 'KENNEDY', 'USAQUEN'],
    'codigo_divipola': ['1100107', '1100119', '1100105', '1100108', '1100101'],
    'ipt': [85.4, 82.1, 78.9, 65.2, 34.0]
})

df_ranking = viz.build_ranking(df_ejemplo_ipt, score_column='ipt')
df_ranking['posicion_ranking'] = range(1, len(df_ranking) + 1)
display(df_ranking)
```

## 3. Exportación a `data/curated`

```python
out_dashboard = viz.export_for_dashboard(df_ranking, 'dashboard_ranking_muestra.csv')
print(f"Archivo para dashboard exportado a: {out_dashboard}")
```
