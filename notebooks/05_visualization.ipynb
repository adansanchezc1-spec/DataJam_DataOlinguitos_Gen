## Objetivos
- Cargar resultados curados y preparar tablas para visualización.- Generar mapas y ranking por localidad.- Exportar datos en formatos compatibles con el dashboard.

```python
import pandas as pd
from pathlib import Path

ROOT = Path('..').resolve()
CURATED_DIR = ROOT / 'data' / 'curated'
CURATED_DIR.mkdir(parents=True, exist_ok=True)

def load_curated(filename: str) -> pd.DataFrame:
    path = CURATED_DIR / filename
    assert path.exists(), f'No existe {path}'
    return pd.read_csv(path)

# df = load_curated('results.csv')
```

## Preparar ranking por localidad

```python
def build_ranking(df: pd.DataFrame, score_col: str = 'ipt') -> pd.DataFrame:
    return df.sort_values(by=score_col, ascending=False).reset_index(drop=True)

# ranking = build_ranking(df)
# ranking.head()
```

## Exportar para dashboard

```python
def export_for_dashboard(df: pd.DataFrame, filename: str) -> Path:
    path = CURATED_DIR / filename
    df.to_csv(path, index=False)
    return path

# export_for_dashboard(ranking, 'dashboard_ranking.csv')
```
